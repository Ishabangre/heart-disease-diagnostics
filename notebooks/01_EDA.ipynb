# 01 - Exploratory Data Analysis

```
import sys
sys.path.append('..')   # so `src` can be imported
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from src.preprocessing import load_data
df = load_data('../dataset/heart.csv')
df.head()
```

```
df.info()
df.describe()
```

```
print('Missing values:', df.isnull().sum().sum())
print('Duplicate rows:', df.duplicated().sum())
df['target'].value_counts()
```

```
sns.countplot(x='target', data=df)
plt.title('Target distribution')
plt.savefig('../outputs/graphs/target_distribution.png', dpi=150)
plt.show()
```

```
plt.figure(figsize=(11,8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation heatmap')
plt.savefig('../outputs/graphs/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```
num_cols = ['age','trestbps','chol','thalach','oldpeak']
df[num_cols].hist(bins=20, figsize=(12,7))
plt.tight_layout()
plt.savefig('../outputs/graphs/numeric_distributions.png', dpi=150)
plt.show()
```

```
fig, axes = plt.subplots(1, 3, figsize=(15,4))
for ax, c in zip(axes, ['sex','cp','exang']):
    sns.countplot(x=c, hue='target', data=df, ax=ax)
plt.tight_layout()
plt.show()
```
