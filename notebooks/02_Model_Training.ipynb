# 02 - Model Training

```
import sys
sys.path.append('..')   # so `src` can be imported
from src.preprocessing import *
from src.feature_engineering import add_features
from src.model import build_model, AVAILABLE_MODELS
from sklearn.model_selection import cross_val_score

df = add_features(clean_data(load_data('../dataset/heart.csv')))
X, y = split_features_target(df)
X_train, X_test, y_train, y_test = split_data(X, y)
print(X.shape, X_train.shape)
```

```
for name in AVAILABLE_MODELS:
    s = cross_val_score(build_model(name), X_train, y_train, cv=5)
    print(f'{name:22s} CV acc = {s.mean():.4f} (+/- {s.std():.4f})')
```

```
model = build_model('logistic_regression').fit(X_train, y_train)
print('Train acc:', model.score(X_train, y_train))
print('Test  acc:', model.score(X_test, y_test))
```

To save the final model + metrics, run `python -m src.train` from the project root.
